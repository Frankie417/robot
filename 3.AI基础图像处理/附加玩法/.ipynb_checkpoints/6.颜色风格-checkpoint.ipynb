# 6.颜色风格

@  文件名：6，颜色风格.ipynb <br>
@  版本：V2.0 <br>
@  说明：通过OpenCV改变图片的颜色风格<br>
我们来学习一下OpenCV课程！！<br>

```python
import cv2
import numpy as np

img = cv2.imread('image0.jpg',1)
#cv2.imshow('src',img)

imgInfo = img.shape
height = imgInfo[0]
width = imgInfo[1]
#rgb -》RGB new “蓝色”
# b=b*1.5
# g = g*1.3
dst = np.zeros((height, width, 3), np.uint8)

for i in range(0,height):
    for j in range(0,width):
        (b,g,r) = img[i,j]
        b = b*1.5
        g = g*1.3
        if b>255:
            b = 255
        if g>255:
            g = 255
        dst[i,j]=(b,g,r)
        
# cv2.imshow('dst',dst)
# cv2.waitKey(0)
```

# 原始图像

```python
import matplotlib.pyplot as plt

img_bgr2rgb1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_bgr2rgb1)
plt.show()
```

# 颜色风格

```python
img_bgr2rgb1 = cv2.cvtColor(dst, cv2.COLOR_BGR2RGB)
plt.imshow(img_bgr2rgb1)
plt.show()
```
