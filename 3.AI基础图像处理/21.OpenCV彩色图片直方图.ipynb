# 21.OpenCV彩色图片直方图

@  文件名：20.OpenCV彩色图片直方图.py <br>
@  版本：V2.0 <br>
@  说明：通过OpenCV计算图片直方图<br>
我们来学习一下OpenCV课程！！<br>

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('./images/dice007.jpg', 1)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()
```

# plot绘制直方图

```python
plt.hist(img.ravel(), 256)   #ravel() 二维降一维  256灰度级的分组情况
plt.show()
```

```python
# plot绘制opencv返回的直方图值
```

```python
histb = cv2.calcHist([img], [0], None, [256], [0, 255])
plt.plot(histb, color='b')
plt.show()
```
