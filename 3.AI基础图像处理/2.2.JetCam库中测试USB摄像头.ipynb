### 载入对应的库文件

```python
from jetcam.usb_camera import USBCamera

camera = USBCamera(width=640, height=480, capture_width=640, capture_height=480, capture_device=0)
#camera = USBCamera(capture_device=0)
```

捕获一帧图片

```python
image = camera.read()

print(image.shape)
```

调用 ``read`` 同时会更新对应的值 ``value``

```python
print(camera.value.shape)
```

您可以创建一个小部件来显示这个图像。您需要将bgr8格式转换为jpeg格式，再将流转换为浏览器格式

```python
import ipywidgets
from IPython.display import display
from jetcam.utils import bgr8_to_jpeg

image_widget = ipywidgets.Image(format='jpeg')

image_widget.value = bgr8_to_jpeg(image)

display(image_widget)
```

你可以设置相机的 ``running`` 值，在后台持续更新该值。这允许你附加回调相机值的变化。

```python
camera.running = True

def update_image(change):
    image = change['new']
    image_widget.value = bgr8_to_jpeg(image)
    
camera.observe(update_image, names='value')
```

您可以像这样取消这个回调

```python
camera.unobserve(update_image, names='value')
```

你也可以使用traitlets  ``dlink `` 方法来连接摄像头到小部件，使用一个转换

```python
import traitlets

camera_link = traitlets.dlink((camera, 'value'), (image_widget, 'value'), transform=bgr8_to_jpeg)
```

您可以像这样删除这个链接

```python
camera_link.unlink()
```

像这样重新连接

```python
camera_link.link()
```
