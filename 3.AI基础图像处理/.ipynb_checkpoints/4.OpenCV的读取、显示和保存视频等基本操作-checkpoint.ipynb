## 1、用摄像头捕获视频

```python
# 载入库
import cv2
from jetcam.utils import bgr8_to_jpeg
# 线程函数操作库
import threading # 线程
import ctypes
import inspect
#bgr8转jpeg格式
import enum
#import cv2

def bgr8_to_jpeg(value, quality=75):
    return bytes(cv2.imencode('.jpg', value)[1])
```

### 1.1  创建显示控件

```python
import cv2
import ipywidgets.widgets as widgets
import threading
import time
import traitlets
from IPython.display import display

image_widget = widgets.Image(format='jpeg', width=600, height=500)  #设置摄像头显示组件
display(image_widget)   
```

### 1.2 线程的结束代码

```python
def _async_raise(tid, exctype):
    tid = ctypes.c_long(tid)
    if not inspect.isclass(exctype):
        exctype = type(exctype)
    res = ctypes.pythonapi.PyThreadState_SetAsyncExc(tid, ctypes.py_object(exctype))
    if res == 0:
        raise ValueError("invalid thread id")
    elif res != 1:
        ctypes.pythonapi.PyThreadState_SetAsyncExc(tid, None)
        raise SystemError("PyThreadState_SetAsyncExc failed")
        
def stop_thread(thread):
    _async_raise(thread.ident, SystemExit)
```

### 1.3 动态显示摄像头视频

```python
image = cv2.VideoCapture(0)                           #打开摄像头

width=600
height=500

ret, frame = image.read()     #读取摄像头数据
image_widget.value = bgr8_to_jpeg(frame)

def Video_display():
    while True:
        ret, frame = image.read()
        image_widget.value = bgr8_to_jpeg(frame)
    # 
    cap.release()
    
t = threading.Thread(target=Video_display)
t.setDaemon(True)
t.start()
```

## 2、从文件中播放视频

```python
# -*- coding: utf-8 -*-
from jetcam.utils import bgr8_to_jpeg
import numpy as np
import cv2
```

```python
import traitlets
import ipywidgets.widgets as widgets
from IPython.display import display
face_image = widgets.Image(format='jpeg', width=640, height=480)
display(face_image)
```

```python
cap = cv2.VideoCapture('./images/walking.avi')

while cap.isOpened():
    ret, frame = cap.read()
    # 如果正确读取帧，ret为True
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break
    frame = cv2.flip(frame,4)
    face_image.value = bgr8_to_jpeg(frame)

```

## 3.保存视频

```python
# 载入库
import cv2
from jetcam.utils import bgr8_to_jpeg
# 线程函数操作库
import threading # 线程
import ctypes
import inspect
#bgr8转jpeg格式
import enum
#import cv2

def bgr8_to_jpeg(value, quality=75):
    return bytes(cv2.imencode('.jpg', value)[1])
```

```python
import cv2
import ipywidgets.widgets as widgets
import threading
import time
import traitlets
from IPython.display import display

image_widget = widgets.Image(format='jpeg', width=600, height=500)  #设置摄像头显示组件
display(image_widget)   
```

```python
image = cv2.VideoCapture(0)                           #打开摄像头

width=600
height=500

ret, frame = image.read()     #读取摄像头数据
image_widget.value = bgr8_to_jpeg(frame)
```

```python
# 定义编解码器并创建VideoWriter对象
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter('output.avi', fourcc, 20.0, (640, 480))
while image.isOpened():
    ret, frame = image.read()
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break
    frame = cv2.flip(frame, 4)
    # 写翻转的框架
    out.write(frame)
    image_widget.value = bgr8_to_jpeg(frame)
    
# 完成工作后释放所有内容
cap.release()
out.release()
```
