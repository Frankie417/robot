```python
import time
import Jetson.GPIO as GPIO
import Adafruit_SSD1306

from PIL import Image
from PIL import ImageDraw
from PIL import ImageFont


# =========================
# 1. 创建 OLED 对象
# =========================

disp = Adafruit_SSD1306.SSD1306_128_32(
    rst=None,
    i2c_bus=1,
    gpio=GPIO
)


# =========================
# 2. 初始化 OLED
# =========================

disp.begin()

disp.clear()
disp.display()


# =========================
# 3. 创建画布
# =========================

width = disp.width
height = disp.height

image = Image.new(
    '1',
    (width, height)
)

draw = ImageDraw.Draw(image)

font = ImageFont.load_default()


# =========================
# 4. 画文字
# =========================

draw.rectangle(
    (0, 0, width, height),
    outline=0,
    fill=0
)

draw.text(
    (0, 0),
    "DICE AI arm5dof",
    font=font,
    fill=255
)


# =========================
# 5. OLED 显示
# =========================

disp.image(image)
disp.display()
```
