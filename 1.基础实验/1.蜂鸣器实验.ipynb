### 1.导入必要的库文件(运行程序前，请先关闭其他程序内核)

```python
# 导入Jetson Nano  GPIO库
import Jetson.GPIO as GPIO
# 从time模块导入sleep函数
from time import sleep
# 暂时忽略警告
GPIO.setwarnings(False)
# 使用实际的PIN管脚编码
GPIO.setmode(GPIO.BOARD) 
# 设置蜂鸣器-引脚12作为输出
whitebear_buzzer=11

# 将蜂鸣器引脚设置为输出引脚，并将初始值设置为LOW（关闭）
GPIO.setup(whitebear_buzzer, GPIO.OUT, initial=GPIO.LOW) 
```

### 2.运行10次

```python
for i in range(0,10):
    GPIO.output(whitebear_buzzer,GPIO.HIGH)
    print ("Beep")
    sleep(0.5) # 延时0.5s
    GPIO.output(whitebear_buzzer,GPIO.LOW)
    print ("No Beep")
    sleep(0.5)  
```
