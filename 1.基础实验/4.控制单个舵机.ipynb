### 1.导入必要的库文件(运行程序前，请先关闭其他程序内核)

```python
# 导入依赖
import time
import struct
import serial
from uservo import UartServoManager
```

## 2、设置串口总线参数

```python
DEVICE_PORT_DEFAULT = '/dev/ttyTHS1'   # Jetson Nano 
SERVO_BAUDRATE = 115200                # 舵机的波特率
JOINT1_SERVO_ID = 0                    # 关节1
JOINT2_SERVO_ID = 1                    # 关节2
JOINT3_SERVO_ID = 2                    # 关节3
JOINT4_SERVO_ID = 3                    # 关节4
JOINT5_SERVO_ID = 4                    # 关节5
JOINT6_SERVO_ID = 5                    # 关节6
```

### 3、初始化舵机管理器

```python
uart = serial.Serial(port=DEVICE_PORT_DEFAULT, baudrate=SERVO_BAUDRATE,parity=serial.PARITY_NONE, stopbits=1,bytesize=8,timeout=0)
uservo = UartServoManager(uart, is_debug=True)
```

### 4、控制舵机旋转

```python
print("设置平移舵机角度为0.0°,设置转速为200 °/s, 加速时间100ms, 减速时间100ms")
uservo.set_servo_angle(JOINT6_SERVO_ID, 30.0, velocity=200.0, t_acc=100, t_dec=100) # 设置舵机角度 
uservo.wait() # 等待舵机静止
```
