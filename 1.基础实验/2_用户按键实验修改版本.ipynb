```python
# 导入 Jetson Nano GPIO 库
import Jetson.GPIO as GPIO
from time import sleep

# -------------------------
# 1. 定义引脚
# -------------------------
whitebear_button_pin1 = 37
whitebear_button_pin2 = 40
whitebear_buzzer_pin = 11

# -------------------------
# 2. GPIO 初始化
# -------------------------

# 关闭警告
GPIO.setwarnings(False)

# 使用物理引脚编号
GPIO.setmode(GPIO.BOARD)

# 两个按键作为输入，并启用上拉
GPIO.setup(
    whitebear_button_pin1,
    GPIO.IN,
    pull_up_down=GPIO.PUD_UP
)

GPIO.setup(
    whitebear_button_pin2,
    GPIO.IN,
    pull_up_down=GPIO.PUD_UP
)

# 蜂鸣器作为输出，初始关闭
GPIO.setup(
    whitebear_buzzer_pin,
    GPIO.OUT,
    initial=GPIO.LOW
)

# -------------------------
# 3. 执行检测
# -------------------------
try:
    while True:

        # 读取两个按钮
        ver1 = GPIO.input(whitebear_button_pin1)
        ver2 = GPIO.input(whitebear_button_pin2)

        # 检查按键是否按下
        if ver1 == 0 or ver2 == 0:
            # 有一个按钮按下，蜂鸣器响
            GPIO.output(whitebear_buzzer_pin, GPIO.HIGH)

        else:
            # 没有按钮按下，关闭蜂鸣器
            GPIO.output(whitebear_buzzer_pin, GPIO.LOW)

        sleep(0.01)

except KeyboardInterrupt:
    print("程序停止")

finally:
    GPIO.output(whitebear_buzzer_pin, GPIO.LOW)
    GPIO.cleanup()
```
