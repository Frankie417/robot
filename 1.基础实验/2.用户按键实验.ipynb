### 1.导入必要的库文件(运行程序前，请先关闭其他程序内核)

```python
# 导入Jetson Nano  GPIO库
import Jetson.GPIO as GPIO
from time import sleep
```

### 2. 配置管脚

```python
# 配置按钮和蜂鸣器引脚
whitebear_button_pin1 = 37 # 按键key1
whitebear_button_pin2 = 40 # 按键key2

whitebear_buzzer_pin = 11 # 蜂鸣器
# 暂时忽略警告
GPIO.setwarnings(False)
# 使用实际的PIN管脚编码
GPIO.setmode(GPIO.BOARD)


# 设置按钮引脚按键管脚为输入和蜂鸣器引脚作为输出
GPIO.setup(whitebear_button_pin1, GPIO.IN, pull_up_down=GPIO.PUD_UP)
GPIO.setup(whitebear_button_pin2, GPIO.IN, pull_up_down=GPIO.PUD_UP)
# 将蜂鸣器引脚设置为输出引脚，并将初始值设置为LOW（关闭）
GPIO.setup(whitebear_buzzer_pin, GPIO.OUT,initial=GPIO.LOW)
```

### 3.执行检测

```python
while True:
    #GPIO.setup(whitebear_button_pin1, GPIO.IN)
    ver1 = GPIO.input(whitebear_button_pin1)
    ver2 = GPIO.input(whitebear_button_pin2)
    #GPIO.setup(whitebear_button_pin, GPIO.OUT,initial=GPIO.HIGH)
    # 检查按键是否按下
    if(ver1 ==0 or ver2 ==0):
        GPIO.output(whitebear_buzzer_pin, GPIO.HIGH)
    else:
        # 没有按键按下, 关闭蜂鸣器
        GPIO.output(whitebear_buzzer_pin, GPIO.LOW)
```
