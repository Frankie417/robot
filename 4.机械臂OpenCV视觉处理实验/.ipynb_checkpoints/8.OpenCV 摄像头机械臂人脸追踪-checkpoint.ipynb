```python
# 线程函数操作库
import threading # 线程
import ctypes
import inspect

# 线程结束代码
def _async_raise(tid, exctype):
    tid = ctypes.c_long(tid)
    if not inspect.isclass(exctype):
        exctype = type(exctype)
    res = ctypes.pythonapi.PyThreadState_SetAsyncExc(tid, ctypes.py_object(exctype))
    if res == 0:
        raise ValueError("invalid thread id")
    elif res != 1:
        ctypes.pythonapi.PyThreadState_SetAsyncExc(tid, None)
        raise SystemError("PyThreadState_SetAsyncExc failed")
        
def stop_thread(thread):
    _async_raise(thread.ident, SystemExit)
```

```python
import traitlets
import ipywidgets.widgets as widgets
from IPython.display import display
from jetcam.utils import bgr8_to_jpeg


frame_img = widgets.Image(format='jpeg', width=640, height=480)
display(frame_img)
```

```python
# 载入人脸和眼睛的HAAR 模型
face_cascade = cv2.CascadeClassifier('./images/haarcascade_frontalface_default.xml')
eye_cascade = cv2.CascadeClassifier('./images/haarcascade_eye.xml')
```

```python
def Video_display():
    global pan_angle
    global tilt_angle
    
    capture = cv2.VideoCapture(0)
    
    dispW=640
    dispH=480
    
    while True: 
        ret, cframe_data = capture.read()
        if not ret:
            return ret, cframe_data
        cframe_data=cframe_data.copy()
        
        gray = cv2.cvtColor(cframe_data,cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray,1.3,5)

        for(x,y,w,h) in faces:
            cv2.rectangle(cframe_data,(x,y),(x+w,y+h),(0,0,255),2)
            Xcent = x + w/2
            Ycent = y + h/2
            
            errorPan  = Xcent - dispW/2 
            errorTilt = Ycent - dispH/2
            
            if abs(errorPan)>15:
                pan_angle=pan_angle+errorPan/75
            if abs(errorTilt)>15:
                tilt_angle=tilt_angle-errorTilt/75
                
            if pan_angle > 90:
                pan = 90
                print("Pan out of Range")
            if pan_angle < -90:
                pan_angle = -90
                print("pan Out of Range")
            if tilt_angle > 90:
                tilt_angle = 90
                print("Pan out of Range")
            if tilt_angle < -90:
                tilt_angle = -90
                print("pan Out of Range")
                
            arm.uservo.set_servo_angle(0,pan_angle,interval=0)
            arm.uservo.set_servo_angle(2,tilt_angle,interval=0)

            roi_gray = gray[y:y+h, x:x+w]
            roi_color = cframe_data[y:y+h, x:x+w]        
            eyes = eye_cascade.detectMultiScale(roi_gray)
            for (ex,ey,ew,eh) in eyes:
                cv2.rectangle(roi_color,(ex,ey),(ex+ew,ey+eh),(0,255,0),2)     
                
        frame_img.value = bgr8_to_jpeg(cframe_data)
        
    capture.release()
```

```python
# 开始线程
t = threading.Thread(target=Video_display)
t.setDaemon(True)
t.start()
```

```python
# 结束线程
stop_thread(t)
```
