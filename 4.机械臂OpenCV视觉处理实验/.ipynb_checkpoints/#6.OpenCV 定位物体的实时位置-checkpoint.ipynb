```python
# 载入必要的库文件
from __future__ import print_function
import numpy as np
from imutils.video import VideoStream
import imutils
import time
import cv2
import os
import socket
```

```python
# 打印物体的实时位置
def mapObjectPosition (x, y):
    print ("[INFO] Object Center coordenates at X0 = {0} and Y0 =  {1}".format(x, y))
```

```python
# 定义对象的上下边界
# 在HSV颜色空间中进行跟踪
colorLower = (9,135,231)
colorUpper = (31,255,255)
```

```python
# 创建显示控件
import traitlets
import ipywidgets.widgets as widgets
from IPython.display import display
from jetcam.utils import bgr8_to_jpeg

Frame = widgets.Image(format='jpeg', width=500, height=350)
display(Frame)
```

```python
# 线程函数操作库
import threading # 线程
import ctypes
import inspect

# 线程结束代码
def _async_raise(tid, exctype):
    tid = ctypes.c_long(tid)
    if not inspect.isclass(exctype):
        exctype = type(exctype)
    res = ctypes.pythonapi.PyThreadState_SetAsyncExc(tid, ctypes.py_object(exctype))
    if res == 0:
        raise ValueError("invalid thread id")
    elif res != 1:
        ctypes.pythonapi.PyThreadState_SetAsyncExc(tid, None)
        raise SystemError("PyThreadState_SetAsyncExc failed")
        
def stop_thread(thread):
    _async_raise(thread.ident, SystemExit)
```

```python
def Video_display():
    capture = cv2.VideoCapture(0)
    # 循环的帧从视频流
    while True:
        # 从视频流中抓取下一帧，调整大小
        # 帧，并将其转换为HSV颜色空间
        ret, cframe_data = capture.read()
        if not ret:
            return ret, cframe_data
        
        cframe_data = imutils.resize(cframe_data, width=500)
        cframe_data = imutils.rotate(cframe_data, angle=0)
        hsv = cv2.cvtColor(cframe_data, cv2.COLOR_BGR2HSV)

        # 为对象颜色构造一个遮罩，然后执行
        # 一系列的膨胀和侵蚀，以消除任何小的
        # blobs left in the mask
        mask = cv2.inRange(hsv, colorLower, colorUpper)
        mask = cv2.erode(mask, None, iterations=2)
        mask = cv2.dilate(mask, None, iterations=2)

        # 找到遮罩中的轮廓并初始化
        # (x, y) center of the object
        cnts = cv2.findContours(mask.copy(), cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)[-2]
        #cnts = cnts[0] if imutils.is_cv2() else cnts[1]
        center = None

        # 只有在找到至少一条轮廓线时才进行
        if len(cnts) > 0:
            # 在蒙版中找到最大的轮廓，然后使用
            # 它可以计算出最小的围圆
            # 重心
            c = max(cnts, key=cv2.contourArea)
            ((x, y), radius) = cv2.minEnclosingCircle(c)
            M = cv2.moments(c)
            center = (int(M["m10"] / M["m00"]), int(M["m01"] / M["m00"]))

            # 只有当半径满足最小尺寸时才进行
            if radius > 10:
                # 在框架上画圆和质心，
                # 然后更新跟踪点的列表
                cv2.circle(cframe_data, (int(x), int(y)), int(radius),(0, 255, 255), 2)
                cv2.circle(cframe_data, center, 5, (0, 0, 255), -1)

                # 定位舵机在圆心
                mapObjectPosition(int(x), int(y))
        # 向我们的屏幕显示框架
        Frame.value = bgr8_to_jpeg(cframe_data)    
    # 做点清理工作
    print("\n [INFO] Exiting Program and cleanup stuff \n")
    capture.release()
```

```python
# 开始线程
t = threading.Thread(target=Video_display)
t.setDaemon(True)
t.start()
```

```python
# 结束线程
stop_thread(t)
```
