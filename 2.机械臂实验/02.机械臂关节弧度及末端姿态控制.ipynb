###  1.导入依赖(运行程序前，请先关闭其他程序内核)

```python
import math
from arm5dof_uservo import Arm5DoFUServo
from config import *
```

### 2.机械臂创建

```python
arm = Arm5DoFUServo()
```

```python
arm.home()
```

### 3.机械臂关节控制

```python
arm.set_joint_angle_list([0, -math.pi/2, 0, 0, 0, math.pi/4], is_wait=True)
```

```python
arm.set_joint_angle_list([0, -3*math.pi/4, math.pi/3, math.pi/2, 0, 0], is_wait=True)
```

#### 3.1关节控制Minimum Jerk轨迹规划

```python
arm.set_joint_angle_list_soft([math.pi/2, -math.pi/2, math.pi/4, 0], T=1.5)
arm.set_joint_angle_list_soft([-math.pi/2, -math.pi/2, math.pi/3, math.pi/3], T=2.0)
arm.set_joint_angle_list_soft([0, -3*math.pi/4, math.pi/2, math.pi/3], T=1.5)
arm.set_joint_angle_list_soft([0, -math.pi/2, 0, 0], T=1.5)
arm.set_joint_angle_list_soft([0, -np.pi/3, 0, 0], T=1.5)
arm.set_joint_angle_list_soft([math.pi/2, -math.pi/2, 0, 0], T=1.5)
arm.home()
print("完成动作")
```

### 4.机械臂末端位资控制

#### 4.1、机械臂末端控制(自动生成Pitch)

```python
p_tool = [150.0, 80.0, 200.0]
arm.set_tool_pose(p_tool)
```

```python
p_tool = [200.0, -40.0, 100.0]
arm.set_tool_pose(p_tool)
```

```python
p_tool = [140.0, 0.0, 150.0]
tool_roll = np.radians(-135.0)
arm.set_tool_pose(p_tool, tool_roll=tool_roll)
```

```python
p_tool = [180.0, -50.0, 100.0]
tool_roll = np.radians(90.0)
arm.set_tool_pose(p_tool, tool_roll=tool_roll)
```

### 4.1、末端控制(设定Pitch)

```python
tool_posi = [180.0, -50.0, 100.0]
tool_pitch = np.radians(110.0)
tool_roll = np.radians(180.0)
arm.set_tool_pose(tool_posi, tool_pitch=tool_pitch, tool_roll=tool_roll)
```

### 4.2、获取当前末端位姿

```python
arm.set_damping(1000) 
```

```python
# 返回的数据格式: [x6, y6, z6, pitch, roll]  
x, y, z, pitch, roll = arm.get_tool_pose()
print(f"坐标: X {x:.1f}  Y {y:.1f} Z {z:.1f}")
print(f"俯仰角(Pitch) {np.degrees(pitch):.1f} deg")
print(f"横滚角(Roll) {np.degrees(roll):.1f} deg")
```
