### 3.机械臂舞蹈动作

```python
# 回到Home 点
arm.home()

# 位置1
tool_posi = [180.0, -50.0, 100.0]
tool_pitch = np.radians(110.0)
tool_roll = np.radians(180.0)
arm.set_tool_pose(tool_posi, tool_pitch=tool_pitch, tool_roll=tool_roll)

# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()

# 位置2
tool_posi = [180.0, 50.0, 100.0]
tool_pitch = np.radians(110.0)
tool_roll = np.radians(180.0)
arm.set_tool_pose(tool_posi, tool_pitch=tool_pitch, tool_roll=tool_roll)

# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()

# 回到Home 点
arm.home()


# 位置3
tool_posi = [180.0, -50.0, 110.0]
tool_pitch = np.radians(110.0)
tool_roll = np.radians(180.0)
arm.set_tool_pose(tool_posi, tool_pitch=tool_pitch, tool_roll=tool_roll)

# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()

# 位置4
tool_posi = [180.0, 50.0, 110.0]
tool_pitch = np.radians(110.0)
tool_roll = np.radians(180.0)
arm.set_tool_pose(tool_posi, tool_pitch=tool_pitch, tool_roll=tool_roll)

# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()
# 爪子打开
arm.gripper_open()
# 爪子闭合
arm.gripper_close()

# 回到Home 点
arm.home()
```
