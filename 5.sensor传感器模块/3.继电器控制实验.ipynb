# 3.继电器控制实验

@  说明：继电器控制实验
从前面的LED灯的控制，我们来学习比较常用的继电器控制实验，可以通过小电流控制大电流设备！<br>

## 3.1 导入必要的库文件

```python
import RPi.GPIO as GPIO
import time
```

## 3.2 定义继电器的GPIO口

```python
RelayPin = 29    # 定义继电器管脚为Pin11
```

## 3.3 初始化工作

```python
# 初始化工作
def setup():
    GPIO.setmode(GPIO.BOARD)        # 采用实际的物理管脚给GPIO口
    GPIO.setwarnings(False)        #  忽略GPIO警告
    GPIO.setup(RelayPin, GPIO.OUT)  # 设置Pin模式为输出模式
    GPIO.output(RelayPin, GPIO.LOW) # 关闭继电器
```

## 3.4 循环执行

```python
# 循环函数
def loop():
    while True:
        # 如果是低电平有效，则为关闭继电器
        # 继电器打开
        GPIO.output(RelayPin, GPIO.HIGH)
        time.sleep(0.5)                      # 延时500ms
        # 如果是低电平有效，则为使能继电器
        # 继电器关闭
        GPIO.output(RelayPin, GPIO.LOW)
        time.sleep(0.5) 
```

## 3.5 程序入口执行

```python
# 释放资源
def destroy():
    GPIO.output(RelayPin, GPIO.LOW) # 关闭继电器
    GPIO.cleanup()                      # 释放资源

# 程序入口
if __name__ == '__main__':
    setup()           #  初始化
    try:
        loop()        #  调用循环函数
    except KeyboardInterrupt:  #  当按下Ctrl+C时，将执行destroy()子程序。
        destroy()     #  释放资源
```
