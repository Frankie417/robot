# 7.霍尔传感器实验

@  文件名：7.霍尔传感器实验.py <br>
@  版本：V2.0 <br>
@  说明：模拟霍尔传感器实验
我们一起来学习一下模拟霍尔传感器实验吧！<br>

```python
import RPi.GPIO as GPIO
import PCF8591 as ADC
import time

# 模块地址设置
def setup():
    ADC.setup(0x48) # 设置PCF8591模块地址

# 打印出磁场的信息
def Print(x):
    if x == 0:    #  没有磁场
        print ('')
        print ('*************')
        print ('* No Magnet *')
        print ('*************')
        print ('')
    if x == 1:    #  磁场为南
        print ('')
        print ('****************')
        print ('* Magnet North *')
        print ('****************')
        print ('')
    if x == -1:   # 磁场为北
        print ('')
        print ('****************')
        print ('* Magnet South *')
        print ('****************')
        print ('')

# 无限循环
def loop():
    status = 0 # 给状态变量赋0值
    while True:  # 无限循环
        res = ADC.read(0) # 获取AIN0 的模拟量值；
        print ('Current intensity of magnetic field : ', res) # 打印出磁场的电流强度值
        if res - 133 < 5 and res - 133 > -5: # 判断磁场强度范围
            tmp = 0    #  没有磁场
        if res < 128:  # 磁场强度为北
            tmp = -1
        if res > 138:  # 磁场强度为南
            tmp = 1
        if tmp != status: # 磁场发生改变
            Print(tmp) # 调用打印出磁场的信息
            status = tmp # 把当前状态值设置为比较状态值，避免重复打印；
        time.sleep(1.0) # 延时200ms 

# 程序入口
if __name__ == '__main__':
    setup() # 调用初始化程序
    loop()  # 调用循环程序

```
