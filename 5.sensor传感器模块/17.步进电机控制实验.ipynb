# 17.步进电机控制实验

@  文件名：17.步进电机控制实验.py <br>
@  版本：V2.0 <br>
@  说明：步进电机控制实验
我们一起来学习步进电机的驱动方式吧！！<br>

```python
import RPi.GPIO as GPIO
from time import sleep

motorPin = (18,23,24,25)     # 步进电机管脚PIN
rolePerMinute =15            # 每分钟转数
stepsPerRevolution = 2048    # 每转一圈的步数
stepSpeed = (60/rolePerMinute)/stepsPerRevolution  # 每一步所用的时间

# 初始化设置
def setup():
    GPIO.setmode(GPIO.BCM)  # 将GPIO模式设置为BCM编号
    GPIO.setwarnings(False) # 忽略警告
    for i in motorPin:
        GPIO.setup(i, GPIO.OUT) # 设置步进电机的所有管脚为输出模式

# 步进电机旋转
def rotary(clb_direction):
    if(clb_direction == 'a'):     # 逆时针旋转
        for j in range(4):
            for i in range(4):
                GPIO.output(motorPin[i],0x99>>j & (0x08>>i))
            sleep(stepSpeed)

    elif(clb_direction == 'c'):    # 顺时针旋转
        for j in range(4):
            for i in range(4):
                GPIO.output(motorPin[i],0x99<<j & (0x80>>i))
            sleep(stepSpeed)

# 循环函数
def loop():
    while True:
        clb_direction = input('select motor direction a=anticlockwise, c=clockwise: ')
        if(clb_direction == 'c'):
            print('motor running clockwise\n')       # 顺时针旋转
            break
        elif(clb_direction == 'a'):
            print('motor running anti-clockwise\n')  # 逆时针旋转
            break
        else:
            print('input error, please try again!') # 输入错误，再次输入
    while True:
        rotary(clb_direction)       # 让步进电机旋转

# 释放资源
def destroy():
    for i in motorPin:
        GPIO.output(i, GPIO.LOW) # 设置步进电机的所有管脚为输出模式
    GPIO.cleanup() # 释放资源

# 程序入口
if __name__ == '__main__':
    setup()   # 初始化设置函数
    try:
        loop()  # 循环函数
    except KeyboardInterrupt:   # 当按下Ctrl+C时，将执行destroy()子程序。
        destroy()  # 资源释放
```
