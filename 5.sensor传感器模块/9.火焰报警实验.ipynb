```python
import PCF8591 as ADC
import RPi.GPIO as GPIO
import time
import math

whitebear_DO = 16  # 火焰传感器数字IO口
GPIO.setmode(GPIO.BOARD)       # Numbers GPIOs by physical location
GPIO.setwarnings(False)        #  忽略GPIO警告

# 初始化工作
def whitebear_setup():
    ADC.setup(0x48)    # 设置PCF8591模块地址
    GPIO.setup(whitebear_DO, GPIO.IN, pull_up_down=GPIO.PUD_UP) # 设置火焰传感器数字IO口为输入模式

# 打印信息，打印出火焰传感器的状态值
def whitebear_Print(x):
    if x == 1:      # 安全
        print ('')
        print ('   *******************')
        print ('   * whitebear Safe~ *')
        print ('   *******************')
        print ('')
    if x == 0:     # 有火焰
        print ('')
        print ('   ******************')
        print ('   * whitebear Fire! *')
        print ('   ******************')
        print ('')

# 循环函数
def whitebear_loop():
    whitebear_status = 1      # 状态值
    # 无限循环
    while True:
        print (ADC.read(0))  # 读取AIN0的值，打印出火焰传感器值
        # 读取火焰传感器数字IO口
        whitebear_tmp = GPIO.input(whitebear_DO)
        if whitebear_tmp != whitebear_status:     # 判断状态发生改变
            whitebear_Print(whitebear_tmp)        # 打印出火焰传感器的提示信息
            status = whitebear_tmp               # 当前状态值作为下次状态值进行比较，避免重复打印

        time.sleep(1)                         # 延时1s

# 程序入口
if __name__ == '__main__':
    try:
        whitebear_setup() # 初始化
        whitebear_loop()  # 循环函数
    except KeyboardInterrupt: 
        pass
```
