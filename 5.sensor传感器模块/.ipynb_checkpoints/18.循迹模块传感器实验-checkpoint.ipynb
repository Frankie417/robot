# 18.循迹模块传感器实验

@  文件名：18.循迹模块传感器实验.py <br>
@  版本：V2.0 <br>
@  说明：循迹模块传感器实验
我们一起来学习循迹模块的驱动方式吧！！<br>

```python
import RPi.GPIO as GPIO
import time

TrackPin = 16 # 循迹传感器PIN管脚

#  初始化设置函数
def setup():
    GPIO.setmode(GPIO.BOARD)       # 采用实际的物理管脚给GPIO口
    GPIO.setwarnings(False)           # 忽略警告
     # 设置TrackPin管脚为输入模式，上拉至高电平(3.3V)
    GPIO.setup(TrackPin, GPIO.IN, pull_up_down=GPIO.PUD_UP)

# 循环函数
def loop():
    while True:
        if GPIO.input(TrackPin) == GPIO.LOW:  # 检测到白色线
            print ('White line is detected')
        else:
            print ('Black line is detected') # 检测到黑色线

        time.sleep(0.2) # 延时200ms

# 释放资源
def destroy():
    GPIO.cleanup()                     # 释放资源

# 程序入口
if __name__ == '__main__':    
    setup()       # 调用初始化程序
    try:
        loop()    # 调用循环函数
    except KeyboardInterrupt:  #  当按下Ctrl+C时，将执行destroy()子程序。
        destroy()
```
