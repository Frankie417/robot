# 4.U型光电传感器实验

@  文件名：4.U型光电传感器实验.py <br>
@  版本：V2.0 <br>
@  说明：U型光电传感器实验
我们一起来学习一下U型光电传感器实验！<br>

```python
import RPi.GPIO as GPIO
import time

PIPin  = 16  # U型光电传感器管脚定义
Rpin   = 12  # 红色LED模块管脚定义
Gpin   = 13  # 绿色LED模块管脚定义

# GPIO口定义
def setup():
    GPIO.setmode(GPIO.BOARD)       # 采用实际的物理管脚给GPIO口
    GPIO.setwarnings(False)        # 关闭GPIO警告提示
    GPIO.setup(Rpin, GPIO.OUT)     # 设置红色LED管脚为输出模式
    GPIO.setup(Gpin, GPIO.OUT)     # 设置绿色LED管脚为输出模式

    GPIO.setup(PIPin, GPIO.IN, pull_up_down=GPIO.PUD_UP)    # 设置PIPin管脚为输入模式，上拉至高电平(3.3V)
    # 中断函数，当有物体挡住时，调用detect函数
    GPIO.add_event_detect(PIPin, GPIO.BOTH, callback=detect, bouncetime=200)

# 双色LED模块驱动子函数
def double_colorLED(x):
    if x == 0:              # x为1时，开启红色LED灯
        GPIO.output(Rpin, 1)
        GPIO.output(Gpin, 0)
    if x == 1:             # x为0时，开启绿色LED灯
        GPIO.output(Rpin, 0)
        GPIO.output(Gpin, 1)
    time.sleep(0.5)

# 打印函数，显示出是否有黑色物体挡住
def Print(x):
    if x == 0:
        print('*************************************') 
        print('***** Light was blocked!   *') 
        print('*************************************')
    time.sleep(0.5)

# 中断函数，检测到有物体挡住时，响应中断函数
def detect(chn):
    double_colorLED(GPIO.input(PIPin))  # 调用双色LED驱动函数
    Print(GPIO.input(PIPin))   # 打印出检测到有物体挡住！

# 循环函数
def loop():
    while True:
        pass

def destroy():
    GPIO.output(Rpin, GPIO.HIGH)       # 关闭红色LED
    GPIO.output(Gpin, GPIO.HIGH)       # 关闭绿色LED
    GPIO.cleanup()                              # 释放资源

# 程序入口
if __name__ == '__main__':    
    setup() # GPIO 初始化
    try:
        loop()
    except KeyboardInterrupt:  # 当按下Ctrl+C时，将执行destroy()子程序。
        destroy()

```
