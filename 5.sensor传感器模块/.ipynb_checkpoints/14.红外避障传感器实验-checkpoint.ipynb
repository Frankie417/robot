# 14.红外避障传感器实验

@  文件名：14.红外避障传感器实验.py <br>
@  版本：V2.0 <br>
@  说明：红外避障传感器实验
我们一起来学习一下红外避障传感器实验吧！！<br>

```python
import RPi.GPIO as GPIO
import time

ObstaclePin = 16       # 红外避障传感器管脚PIN
GPIO.setwarnings(False)

def setup():
    GPIO.setmode(GPIO.BOARD)       # 采用实际的物理管脚给GPIO口
    GPIO.setup(ObstaclePin, GPIO.IN, pull_up_down=GPIO.PUD_UP) # 设置ObstaclePin管脚为输入模式，上拉至高电平(3.3V)

# 无限循环
def loop():
    while True:
        if (0 == GPIO.input(ObstaclePin)):  # 检测到障碍物
            print ("Detected Barrier!")     # 打印出障碍物信息
            time.sleep(0.2)                          # 延时200ms


# 释放资源
def destroy():
    GPIO.cleanup()                     # 释放资源函数

# 程序入口
if __name__ == '__main__':     
    setup()           # 初始化设置
    try:
        loop()        # 无限循环
    except KeyboardInterrupt:  # 当按下Ctrl+C时，将执行destroy()子程序。
        destroy()              # 调用释放资源
```
