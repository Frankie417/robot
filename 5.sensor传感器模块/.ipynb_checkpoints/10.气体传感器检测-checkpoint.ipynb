# 10.气体传感器检测

@  文件名：10.气体传感器检测.py <br>
@  版本：V2.0 <br>
@  说明：气体传感器检测
我们一起来学习一下气体传感器实验吧！！<br>

```python
import PCF8591 as ADC
import RPi.GPIO as GPIO
import time
import math

whitebear_DO = 23                    # 烟雾传感器数字IO口
whitebear_Buzz = 17                  # led数字IO口
GPIO.setmode(GPIO.BCM)              # 管脚映射，采用BCM编码
GPIO.setwarnings(False)             # 忽略GPIO 警告

# 初始化工作
def whitebear_setup():
    ADC.setup(0x48)                      # 设置PCF8591模块地址
    GPIO.setup(whitebear_DO,GPIO.IN)    # 烟雾传感器数字IO口,设置为输入模式
    GPIO.setup(whitebear_Buzz,GPIO.OUT) # led数字IO口,设置为输出模式
    GPIO.output(whitebear_Buzz,0)        # 设置led为低电平，初始状态关闭led

# 打印信息，打印出是否检测到烟雾信息
def whitebear_Print(x):
    if x == 1:     # 安全
        print ('')
        print ('   ******************')
        print ('   * whitebear Safe~ *')
        print ('   ******************')
        print ('')
    if x == 0:    # 检测到烟雾
        print ('')
        print ('   ************************')
        print ('   * whitebear Danger Gas! *')
        print ('   ************************')
        print ('')

# 循环函数
def whitebear_loop():
    whitebear_status = 1   # 定义状态值变量
    whitebear_count = 0    # 定义计数器变量值
    while True:    # 无限循环
        print (ADC.read(0)) # 获取AIN0 的模拟量值；

        whitebear_tmp = GPIO.input(whitebear_DO)  # 读取GAS烟雾传感器数字IO口值
        if whitebear_tmp != whitebear_status:     # 判断状态发生改变
            whitebear_Print(whitebear_tmp)        # 打印函数，打印出烟雾传感器信息
            whitebear_status = whitebear_tmp      # 把当前状态值设置为比较状态值，避免重复打印；
        if whitebear_status == 0:                # 当检测到烟雾
            whitebear_count += 1                 # 计数器值累计
            # 高低电平交替变化，led闪烁
            if whitebear_count % 2 == 0:         # 进行求余处理，一半为1，一半为0
                GPIO.output(whitebear_Buzz, 1)
            else:
                GPIO.output(whitebear_Buzz, 0)
        else:
            GPIO.output(whitebear_Buzz, 0)       # 设置led为低电平，初始状态关闭led
            whitebear_count = 0                  # 计数器赋0

        time.sleep(1)                         # 延时1s

# 资源释放函数
def destroy():
    GPIO.output(whitebear_Buzz, 0)   # 设置led为低电平，初始状态关闭led
    GPIO.cleanup()         # 释放资源

# 程序入口
if __name__ == '__main__':
    try:
        whitebear_setup()   # 初始化函数
        whitebear_loop()    # 循环函数
    except KeyboardInterrupt: # 当按下Ctrl+C时，将执行destroy()子程序。
        destroy()  #资源释放
```
