# 16.直流电机控制实验

@  文件名：16.直流电机控制实验.py <br>
@  版本：V2.0 <br>
@  说明：直流电机控制实验
我们一起来学习直流电机的驱动方式吧！！<br>

```python
import RPi.GPIO as GPIO
import time


# 设置风扇管脚PIN
MotorPin1   = 23


def setup():
    # 将GPIO模式设置为BCM编号
    GPIO.setmode(GPIO.BCM)
    GPIO.setwarnings(False)           # 忽略警告
    GPIO.setup(MotorPin1, GPIO.OUT)   # 风扇设置为输出模式

# 风扇电机控制函数
def motor(direction):
    # 开启电机
    if direction == 1:
        GPIO.output(MotorPin1, GPIO.HIGH)
        print ("DJ START")

    # 关闭电机
    if direction == 0:
        # 关闭电机
        GPIO.output(MotorPin1, GPIO.LOW)
        print ("DJ STOP")

def main():
    fs_directions = {'Open': 1, 'STOP': 0} # 定义开和关
    while True:
        # 电机开
        motor(fs_directions['Open'])
        time.sleep(5)
        # 电机关
        motor(fs_directions['STOP'])
        time.sleep(5)

# 释放资源
def destroy():
    # 关闭电机
    GPIO.output(MotorPin1, GPIO.LOW)
    # 释放资源
    GPIO.cleanup()    

# 程序入口
if __name__ == '__main__':
    setup()
    try:
        main()
    except KeyboardInterrupt:   # 当按下Ctrl+C时，将执行destroy()子程序。
        destroy()   # 释放资源
```
